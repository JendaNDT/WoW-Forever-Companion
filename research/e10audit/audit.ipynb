# Audit před vydáním

25. 10. 2026. Zrno: druh entity + stabilní ID; polohy se posuzují v konkrétním mapovém rámci. Reprodukce: `python tools/audit_content.py`. Výsledky nejsou potvrzením vydané hry.

```python
import json
from pathlib import Path
root=Path.cwd()
if not (root/'research/e10audit').exists(): root=root.parent.parent
audit=json.loads((root/'research/e10audit/audit-results.json').read_text())
print(json.dumps(audit['counts'],indent=2))
print('Porovnané úkoly:',audit['recorderOverlap'])
print(json.dumps(audit['recorderConflicts'],ensure_ascii=False,indent=2))
print('Ověřená kouzla:',len(audit['spellChecks']))
print('\n'.join(audit['limitations']))
```

Nezáznam není důkaz nedostupnosti. Společný Classic původ není nezávislé potvrzení. Dynamické a herní podmínky zůstávají otevřené.
